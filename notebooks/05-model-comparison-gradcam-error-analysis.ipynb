```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

OUTPUT_DIR = "/kaggle/working/final_outputs"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

print("Notebook 5 Started")
```

```python
comparison_df = pd.DataFrame({

    "Model":[
        "Custom CNN",
        "EfficientNetB0",
        "ResNet50"
    ],

    "Binary Accuracy":[
        0.9518,
        0.9624,
        0.9593
    ],

    "Loss":[
        0.1217,
        0.0977,
        0.1062
    ]
})

comparison_df
```

```python
comparison_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "model_comparison_table.csv"
    ),
    index=False
)

comparison_df
```

```python
plt.figure(figsize=(10,6))

bars = plt.bar(
    comparison_df["Model"],
    comparison_df["Binary Accuracy"]
)

plt.title(
    "Figure 5.1 Model Accuracy Comparison"
)

plt.ylabel("Binary Accuracy")

for bar in bars:

    plt.text(
        bar.get_x()+bar.get_width()/2,
        bar.get_height(),
        round(bar.get_height(),4),
        ha="center"
    )

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "model_accuracy_comparison.pdf"
    )
)

plt.show()
```

```python
plt.figure(figsize=(10,6))

bars = plt.bar(
    comparison_df["Model"],
    comparison_df["Loss"]
)

plt.title(
    "Figure 5.2 Model Loss Comparison"
)

plt.ylabel("Loss")

for bar in bars:

    plt.text(
        bar.get_x()+bar.get_width()/2,
        bar.get_height(),
        round(bar.get_height(),4),
        ha="center"
    )

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "model_loss_comparison.pdf"
    )
)

plt.show()
```

```python
best_model = comparison_df.loc[
    comparison_df["Binary Accuracy"].idxmax()
]

best_model
```

```python
best_model_df = pd.DataFrame(
    best_model
)

best_model_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "best_model.csv"
    )
)

best_model_df
```
